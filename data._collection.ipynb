```python
from nba_api.stats.endpoints import leaguedashplayerstats

stats = leaguedashplayerstats.LeagueDashPlayerStats(season='2024-25')
df_players = stats.get_data_frames()[0]
top_players = df_players.sort_values('FGA', ascending=False).head(100)
player_ids = top_players['PLAYER_ID'].tolist()
```

```python
import time
import pandas as pd
import nba_api.stats.endpoints as endpoints

all_shots = []

for player_id in player_ids:
    try:
        response = endpoints.shotchartdetail.ShotChartDetail(
            team_id = 0,
            player_id = player_id,
            season_nullable = '2024-25',
            context_measure_simple = 'FGA'
        )
        df = response.get_data_frames()[0]
        all_shots.append(df)
        time.sleep(0.6)  # Sleep to avoid hitting rate limits
    except Exception as e:
        print(f"Error fetching data for player ID {player_id}: {e}")

# Combine all shot data into a single DataFrame
full_df = pd.concat(all_shots, ignore_index=True)
full_df.to_csv('nba_shot_data_2024_25.csv', index=False)

```
